```python

import pandas as pd

# Create datetime objects
dates = pd.to_datetime([
    '2026-01-01',
    '2026-01-02',
    '2026-01-03',
    '2026-01-04',
    '2026-01-05'
])

# Create temperature data
temperature = [28, 30, 29, 31, 32]

# Create a time series with timestamps as the index
time_series = pd.Series(
    temperature,
    index=dates
)

# Display the time series
print("Temperature Time Series:")
print(time_series)

# Access temperature for a specific date
print("\nTemperature on 2026-01-03:")
print(time_series.loc['2026-01-03'])

# Calculate daily temperature changes
print("\nDaily Temperature Changes:")
print(time_series.diff())
```
